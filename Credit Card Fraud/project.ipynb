```python
# Fraud by Time Distribution

def time_period(hour):
    if 6 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 18:
        return 'Afternoon'
    elif 18 <= hour < 24:
        return 'Evening'
    else :
        return 'Night'
    
```

```python
# Find Lower and Upper Limits
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
```

```python
print('Lower Limit :',round(lower,2))
print('Upper Limit :',round(upper,2))
```

```python
# detect outlire 
outliers = df[(df['amt']< lower)|(df['amt']>upper)]
```

## Imbalanced Data Handling

```python
df['is_fraud'].value_counts()
```

The dataset is highly imbalanced because the number of genuine transactions is much higher than fraudulent transactions.

```python
# Checking accuracy score
accuracy = accuracy_score(y_pred,y_test)
print('Accuracy :',accuracy)
```

Because the fraud class is very small compared to the genuine class, the model becomes biased toward predicting genuine transactions.

```python
# Appling Random Under Sampling
from imblearn.under_sampling import RandomUnderSampler
rus = RandomUnderSampler(random_state=42)
```

```python
X_under, y_under = rus.fit_resample(X, y)
```

```python
y_under.value_counts()
```

Random Under Sampling reduces the majority class to match the minority class.

```python
# Appling Random Over Sampling
from imblearn.over_sampling import RandomOverSampler
```

```python
ros = RandomOverSampler(random_state=42)
```

```python
X_over, y_over = ros.fit_resample(X, y)
```

```python
y_over.value_counts()
```

Random Over Sampling increases the minority class by duplicating existing samples.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_under,
    y_under,
    test_size=0.2,
    random_state=42
)
```
